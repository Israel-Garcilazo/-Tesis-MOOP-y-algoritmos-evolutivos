**2. Problemas MOP**

Los funciones MOP del libro de Deb. plantean funciones de prueba para calcular frentes de pareto mediante diferentes métodos y algoritmos. A continuación se muestran las funciones en cuestión:

**1. MOP1: problema de Schaffer**

\
  \begin{align}
f_1(x) &= x^2 \\
f_2(x) &= (x - 2)^2
\end{align}
\
Con restricciones:  $-10^5 \le x \le 10^5$.

\
**2. MOP2: función de Fonseca–Fleming**

\
  \begin{align}
f_1(\mathbf{x}) &= 1 - \exp\left(-\sum_{i=1}^{n} \left(x_i - \frac{1}{\sqrt{n}}\right)^2\right) \\
f_2(\mathbf{x}) &= 1 - \exp\left(-\sum_{i=1}^{n} \left(x_i + \frac{1}{\sqrt{n}}\right)^2\right)
\end{align}
\
  Con restricciones: $-4 \le x_i \le 4$, para $i = 1, \dots, n$.

\
**3. MOP3: problema de Poloni**

\
 \begin{align}
f_1(x, y) &= -\left[1 + (A_1 - B_1)^2 + (A_2 - B_2)^2\right] \\
f_2(x, y) &= -\left[(x+3)^2 + (y+1)^2\right]
\end{align}
\
Los escalares se definen de la siguiente manera:

\
\begin{align*}
A_1 &= 0.5 \sin(1) - 2 \cos(1) + \sin(2) - 1.5 \cos(2) \\
A_2 &= 1.5 \sin(1) - \cos(1) + 2 \sin(2) - 0.5 \cos(2) \\
B_1 &= 0.5 \sin(x) - 2 \cos(x) + \sin(y) - 1.5 \cos(y) \\
B_2 &= 1.5 \sin(x) - \cos(x) + 2 \sin(y) - 0.5 \cos(y)
\end{align*}

\
Con reestricciones: $-\pi \le x, y \le \pi$.

\
**4. MOP4: problema de Kursawe**

\
\begin{align}
f_1(\mathbf{x}) &= \sum_{i=1}^{n-1} \left(-10 \exp\left(-0.2 \sqrt{x_i^2 + x_{i+1}^2}\right)\right) \\
f_2(\mathbf{x}) &= \sum_{i=1}^{n} \left(|x_i|^{0.8} + 5 \sin(x_i)^3\right)
\end{align}

\
Con restricciones: $-5 \le x_i \le 5$, para $i = 1, 2, 3$ (usualmente $n=3$).

\
**5. MMOP5: primer problema de Viennet**

\
\begin{align}
f_1(x, y) &= 0.5 (x^2 + y^2) + \sin(x^2 + y^2) \\
f_2(x, y) &= \frac{(3x - 2y + 4)^2}{8} + \frac{(x - y + 1)^2}{27} + 15 \\
f_3(x, y) &= \frac{1}{x^2 + y^2 + 1} - 1.1 \exp(-(x^2 + y^2))
\end{align}

\
Con restricciones: $-30 \le x, y \le 30$.


\
**6. MOP6: problema Discontinuo de Deb**

\
\begin{align}
f_1(x_1, x_2) &= x_1 \\
g(x_1, x_2) &= 1 + 10 x_2 \\
f_2(x_1, x_2) &= g(x_1, x_2) \left[1 - \left(\frac{f_1(x_1, x_2)}{g(x_1, x_2)}\right)^2 - \frac{f_1(x_1, x_2)}{g(x_1, x_2)} \sin(2\pi q f_1(x_1, x_2))\right]
\end{align}

\
Con restricciones: $0 \le x_1, x_2 \le 1$, pero, como se sugiere, podemos fijar $q=4$.

\
**7. MOP7: segundo problema de Viennet**

\
\begin{align}
f_1(x, y) &= \frac{(x-2)^2}{2} + \frac{(y+1)^2}{13} + 3 \\
f_2(x, y) &= \frac{(x+y-3)^2}{36} + \frac{(-x+y+2)^2}{8} - 17 \\
f_3(x, y) &= \frac{(x+2y-1)^2}{175} + \frac{(2y-x)^2}{17} - 13
\end{align}

\
Con restricciones: $-400 \le x, y \le 400$.

**¿Qué enfoque se usará para el algoritmo de solución?**
Como se analizó en el anterior cuaderno de este repositorio, existen algoritmos como el de Kung et al., apodado como *Divide y vencerás*, que sirven eficientemente para disminuir costos computacionales en el ordenamiento de datos y construcción del frente de pareto, pues tiene un costo computacional de $O(N \log N)^{M-2}$, lo que es potencialmente más bajo que los otros dos enfoques cuyos costos son, para ambos casos, de $O(kN^2)$. El pseudocódigo propuesto en el libro de Deb suguiere hacer lo siguiente:

**Paso 1**: se ordena la población de acuerdo al orden descendente de importancia en la primera función objetivo y se renombra la población como P de tamaño N.

**En el Paso 2, Frente(P)**: si |P| = 1, se devuelve P como la salida de la función; de lo contrario, se divide en las mitades:

$T = Frente(P^{(1)} - P^{|P|/2})$ y $B = Frente(P^{(|P|/2 + 1)} - P^{(|P|)})$.

Si la i-ésima solución de B no es dominada por ninguna solución de T, se crea un conjunto fusionado M = T ∪ {i} y se devuelve M como salida de Frente(P), siendo este retorno el conjunto de soluciones no dominadas.

**Puesto en términos sencillos**:

El método ordena todos los puntos de mejor a peor basándose únicamente en la primer función. Esa lista generada la divide a la mitad, creando el grupo superior e inferior.
En el grupo superior, todas las soluciones son no dominadas por el grupo inferior. Entonces se hace pasar un buscador lineal por la segunda función y se descartan aquellos puntos que quedaron en la sección inferior. generando un conjunto de puntos no dominados en la parte superior. Este proceso se repite varias veces hasta que el la parte superior contenga únicamente puntos no dominados para ambas funciones que sean mejores (más grandes en este caso) que todos los puntos de las secciones inferiores, lo que genera el frente de Pareto. Si algun punto no dominado se llega a colar en una sección que pretendía descartarse, entonces se agrega al frente de Pareto como una unión de puntos.

**Esto suguiere que...**

Este método parece ser extremadamente útil para problemas con dos objetivos, pero empieza a tener problemas con funciones de tres o más objetivos. Esto se debe a que el algoritmo va a buscar reducir el problema multiobjetivo a uno de dos objetivos, trabajando unicamente de dos funciones en dos funciones. Es decir, por ejemplo, si se tuvieran tres objetivos $f_1$, $f_2$ y $f_3$, el algoritmo calcularía los frentes de Parteo $FP_1(f_1, f_2)$, $FP_2(f_1, f_3)$ y $FP_3(f_2, f_3)$, y entonces el frente de Pareto del problema sería $FP = FP_1∪FP_2∪FP_3$, lo que lo vuelve demasiado costoso porque estaría haciendo el triple de trabajo de lo que haría en un problema con dos objetivos.

**Lo que significa que, si la teoria no falla, el método de Kung et al. debería de ser más costoso en el problema MOP5 y MOP7 que en el resto de problemas.**

1. Considere los problemas de Optimización multiobjetivo MOP1 a MOP7

Comenzamos con las bibliotecas a usar:

```python
import numpy as np # Biblioteca de cálculo numérico
import matplotlib.pyplot as plt # Biblioteca para graficar
import time # Biblioteca para medir y comparar costos de tiempo.
import pandas as pd # Para la creación de la tabla.
```

Planteamos la configuración general con la que funcionará el algoritmo con todas las funciones:

```python
N_POINTS = 5000  # Esta N se pueden modificar para esperar resultados distintos.

# Creamos la función que va a guardar los puntos generados de forma aleatoria.
def generar_puntos(n_points, bounds, dim):
    # Generamos una matriz aleatoria.
    return np.random.uniform(bounds[0], bounds[1], (n_points, dim))
```

Ahora definimos las funciones de las distintas MOP a continuación:

```python
# Esta es la función que evaluará cada MOP y ordenará los datos en un arreglo matricial.
def eval_mop(name, x):
    X = x[:, 0]
    Y = x[:, 1] if x.shape[1] > 1 else None # Esto sirve para que únicamente se cree si la dimensión es mayor que 1.

    # Definimos de una en una las funciones MOP.
    if name == 'MOP1':
        return np.c_[X**2, (X - 2)**2]

    elif name == 'MOP2':
        n = x.shape[1]
        f1 = 1 - np.exp(-np.sum((x - 1/np.sqrt(n))**2, axis=1))
        f2 = 1 - np.exp(-np.sum((x + 1/np.sqrt(n))**2, axis=1))
        return np.c_[f1, f2]

    elif name == 'MOP3':
        A1, A2 = 0.5*np.sin(1) - 2*np.cos(1) + np.sin(2) - 1.5*np.cos(2), 1.5*np.sin(1) - np.cos(1) + 2*np.sin(2) - 0.5*np.cos(2)
        B1, B2 = 0.5*np.sin(X) - 2*np.cos(X) + np.sin(Y) - 1.5*np.cos(Y), 1.5*np.sin(X) - np.cos(X) + 2*np.sin(Y) - 0.5*np.cos(Y)
        return np.c_[-(1 + (A1 - B1)**2 + (A2 - B2)**2), -((X + 3)**2 + (Y + 1)**2)]

    elif name == 'MOP4':
        f1 = sum(-10 * np.exp(-0.2 * np.sqrt(x[:, i]**2 + x[:, i+1]**2)) for i in range(x.shape[1] - 1))
        f2 = np.sum(np.abs(x)**0.8 + 5 * np.sin(x)**3, axis=1)
        return np.c_[f1, f2]

    elif name == 'MOP5':
        f1, f3 = 0.5*(X**2 + Y**2) + np.sin(X**2 + Y**2), 1/(X**2 + Y**2 + 1) - 1.1*np.exp(-(X**2 + Y**2))
        f2 = ((3*X - 2*Y + 4)**2)/8 + ((X - Y + 1)**2)/27 + 15
        return np.c_[f1, f2, f3]

    elif name == 'MOP6':
        g = 1 + 10 * Y
        f2 = g * (1 - (X/g)**2 - (X/g)*np.sin(2*np.pi * 4 * X))
        return np.c_[X, f2]

    elif name == 'MOP7':
        f1 = ((X - 2)**2)/2 + ((Y + 1)**2)/13 + 3
        f2 = ((X + Y - 3)**2)/36 + ((-X + Y + 2)**2)/8 - 17
        f3 = ((X + 2*Y - 1)**2)/175 + ((2*Y - X)**2)/17 - 13
        return np.c_[f1, f2, f3]
```

Ahora, cada función MOP tiene restricciones y dimensiones diferentes, por ello, no sería práctico suponer trabajar todo en un mismo entorno, pues incluso las escalas y magnitudes de las restricciones son muy distintas entre sí. Para combatir dicho problema, vamos a crear una configuración que guarde las restricciones y dimensiones de cada MOP, a la que podamos llamar más adelante cuando la necesitemos.

```python
mop_config = {
    'MOP1': ([-10**5, 10**5], 1), 'MOP2': ([-4, 4], 3), 'MOP3': ([-np.pi, np.pi], 2),
    'MOP4': ([-5, 5], 3), 'MOP5': ([-30, 30], 2), 'MOP6': ([0, 1], 2), 'MOP7': ([-400, 400], 2)
}
```

A continuación. es turno de implementar el algoritmo de Kung et al.
Aclaremos que:

$P$  es la variable que guarda los vectores ordenados por $f_1(x)$

$F$ es la matriz de los puntos que ya han sido evaluados.

$comparaciones$ es la variable que contiene la lista de vectores que ya han sido comparados y seleccionados.

```python
def kung_et_al(P, F, comparaciones):
    if len(P) <= 1:
        return P  # Caso base

    # Aquí inicia el paso clave: la división recursiva en mitad superior (T) e inferior (B)
    mid = len(P) // 2 # Partimos la lista de vectores a la mitad
    T = kung_et_al(P[:mid], F, comparaciones) # Evaluamos en cada función de las MOP para hallar el intervalo superior.
    B = kung_et_al(P[mid:], F, comparaciones) # Evaluamos en cada función de las MOP para hallar el intervalo inferior.

    M = list(T)  # M contendrá los no dominados combinados

    # Con este ciclo comprobamos si las soluciones de B son dominadas por las de T
    for b in B:
        es_dominado = False
        for t in T:
            comparaciones[0] += 1
            # Con este ciclo checamos que las soluciones de T dominen a las de B, lo que asegura que no se quede algún vector en B que sea no dominado.
            #Recordando que, según el pseudocódigo, T domina a B si es <= en todos y < en al menos uno.
            if np.all(F[t] <= F[b]) and np.any(F[t] < F[b]):
                es_dominado = True
                break
        if not es_dominado:
            M.append(b)

    return M
```

A continuación, generamos las gráficas.

```python
# @title
resultados_tabla = []

for name, (bounds, dim) in mop_config.items():
    # X son los puntos del conjunto y F el mapeo de las funciones objetivos.
    X = generar_puntos(N_POINTS, bounds, dim)
    F = eval_mop(name, X)

    # Aquí definimos esta variable para que, para cada función MOP distinta, se ajuste la escala de graficación.
    indices_ordenados = np.argsort(F[:, 0])

    # Creamos las variables que medirán los costos computacionales.
    comparaciones = [0]
    inicio_tiempo = time.perf_counter()

    # Guardamos las ejecuciones del algoritmo Kung et al. en una variable que toma los puntos no dominados para luego graficar el frente de Pareto.
    frente_indices = kung_et_al(indices_ordenados, F, comparaciones)

    # Creamos el registro de los costos computacionales que se acomodarán en una tabla.
    fin_tiempo = time.perf_counter()
    resultados_tabla.append({
        "Problema": name,
        "Tiempo (s)": fin_tiempo - inicio_tiempo,
        "Comparaciones": comparaciones[0],
        "Puntos Pareto": len(frente_indices)
    })

    # Graficación del dominio y la imagen.
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(f'{name} - Gráfica de Parteo', fontsize=14)

    # Para poder graficar el dominio, usamos unicamente las primeras dos variables.
    x_dim1 = X[:, 0]
    x_dim2 = X[:, 1] if dim > 1 else np.zeros(N_POINTS)
    ax1.scatter(x_dim1, x_dim2, s=2, c='blue', alpha=0.3, label='Dominio')
    ax1.scatter(x_dim1[frente_indices], x_dim2[frente_indices], s=10, c='red', label='Óptimos de Pareto')
    ax1.set_title("Dominio de las funciones")
    ax1.grid(True)

    # Plot del Frente de Pareto, Usando f1 y f2.
    ax2.scatter(F[:, 0], F[:, 1], s=2, c='blue', alpha=0.3, label='Posibles soluciones')
    ax2.scatter(F[frente_indices, 0], F[frente_indices, 1], s=10, c='red', label='Frente de Pareto')
    ax2.set_title("Espacio de los Objetivos")
    ax2.grid(True)

    plt.legend()
    plt.show()
```

Por último, generamos la tabla de comparación de costos mediante el siguiente código.

```python
df_resultados = pd.DataFrame(resultados_tabla)
print("Costos computacionales del aalgoritmo de Kung et al.")
display(df_resultados) # Esta función es, de hecho, una función exclusiva de Jupyter para imprimir DataFrames con buen formato.
```
